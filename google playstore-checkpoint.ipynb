```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("google playstore.csv")

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()
```

```python
missing = df.isnull().sum()

print("Missing Values in Each Column:")
print(missing)

print("\nTotal Missing Values:", missing.sum())
```

```python
# Count duplicate rows
duplicates = df.duplicated().sum()

print("Total Duplicate Rows:", duplicates)
```

```python
# Remove duplicate rows
df = df.drop_duplicates()

# Check the result
print("Dataset Shape After Removing Duplicates:", df.shape)
print("Remaining Duplicate Rows:", df.duplicated().sum())
```

```python
# Fill missing numerical values with the median
for col in df.select_dtypes(include="number").columns:
    df[col] = df[col].fillna(df[col].median())

# Fill missing text values with the mode
for col in df.select_dtypes(include="object").columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])

# Check remaining missing values
print("Total Missing Values After Cleaning:", df.isnull().sum().sum())
```

```python
print("Column Data Types:")
print(df.dtypes)
```

```python
# Clean Reviews column
df["Reviews"] = pd.to_numeric(df["Reviews"], errors="coerce")

# Clean Installs column and convert to numbers
df["Installs"] = (
    df["Installs"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.replace("+", "", regex=False)
)
df["Installs"] = pd.to_numeric(df["Installs"], errors="coerce")

# Clean Price column and convert to numbers
df["Price"] = (
    df["Price"]
    .astype(str)
    .str.replace("$", "", regex=False)
)
df["Price"] = pd.to_numeric(df["Price"], errors="coerce")

print(df[["Reviews", "Installs", "Price"]].dtypes)
```

```python
print("Missing values after type conversion:")
print(df[["Reviews", "Installs", "Price"]].isnull().sum())
```

```python
print(df[["Reviews", "Installs", "Price"]].isnull().sum())
```

```python
# Fill missing values with the median of each column
for col in ["Reviews", "Installs", "Price"]:
    df[col] = df[col].fillna(df[col].median())

# Check the result
print(df[["Reviews", "Installs", "Price"]].isnull().sum())
```

```python
# Select important numerical columns
columns = ["Rating", "Reviews", "Installs", "Price"]

# Create box plots to identify outliers
plt.figure(figsize=(12, 6))

for i, col in enumerate(columns, 1):
    plt.subplot(2, 2, i)
    sns.boxplot(x=df[col])
    plt.title(f"Outlier Detection - {col}")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="Rating", bins=20, kde=True)

plt.title("Distribution of Google Play Store App Ratings")
plt.xlabel("App Rating")
plt.ylabel("Number of Apps")

plt.show()
```

```python
category_rating = df.groupby("Category")["Rating"].mean().sort_values(ascending=False)

plt.figure(figsize=(12, 6))

category_rating.plot(kind="bar")

plt.title("Average App Rating by Category")
plt.xlabel("App Category")
plt.ylabel("Average Rating")
plt.xticks(rotation=90)
plt.tight_layout()

plt.show()
```

```python
# Count apps in each category
category_counts = df["Category"].value_counts().head(10)

plt.figure(figsize=(10, 6))

sns.barplot(x=category_counts.values, y=category_counts.index)

plt.title("Top 10 Categories by Number of Apps")
plt.xlabel("Number of Apps")
plt.ylabel("App Category")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=df,
    x="Price",
    y="Rating",
    alpha=0.5
)

plt.title("Relationship Between App Price and Rating")
plt.xlabel("App Price ($)")
plt.ylabel("App Rating")

plt.tight_layout()
plt.show()
```

```python
# Compare free and paid apps
df["App Type"] = df["Price"].apply(
    lambda price: "Free" if price == 0 else "Paid"
)

app_type_counts = df["App Type"].value_counts()

plt.figure(figsize=(7, 5))
app_type_counts.plot(kind="bar")

plt.title("Number of Free Apps vs Paid Apps")
plt.xlabel("App Type")
plt.ylabel("Number of Apps")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Get the top 10 apps by installs
top_apps = df.nlargest(10, "Installs")

plt.figure(figsize=(10, 6))

sns.barplot(
    data=top_apps,
    x="Installs",
    y="App"
)

plt.title("Top 10 Most Installed Apps")
plt.xlabel("Number of Installs")
plt.ylabel("App Name")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.countplot(
    data=df,
    y="Content Rating",
    order=df["Content Rating"].value_counts().index
)

plt.title("Distribution of Apps by Content Rating")
plt.xlabel("Number of Apps")
plt.ylabel("Content Rating")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=df,
    x="Reviews",
    y="Rating",
    alpha=0.5
)

plt.xscale("log")
plt.title("Relationship Between Reviews and App Ratings")
plt.xlabel("Number of Reviews (Log Scale)")
plt.ylabel("App Rating")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.barplot(data=df, x="App Type", y="Rating", errorbar=None)

plt.title("Average Rating: Free vs Paid Apps")
plt.xlabel("App Type")
plt.ylabel("Average Rating")
plt.ylim(0, 5)

plt.tight_layout()
plt.show()
```

```python
# Select the 10 largest categories
top_categories = df["Category"].value_counts().head(10).index

category_type = df[df["Category"].isin(top_categories)]

plt.figure(figsize=(12, 6))

sns.countplot(
    data=category_type,
    x="Category",
    hue="App Type"
)

plt.title("Free vs Paid Apps Across Top 10 Categories")
plt.xlabel("App Category")
plt.ylabel("Number of Apps")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
# Select numerical columns for correlation
numeric_df = df[["Rating", "Reviews", "Installs", "Price"]]

# Calculate correlation
correlation = numeric_df.corr()

# Draw heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Heatmap of Google Play Store Apps")
plt.tight_layout()
plt.show()
```
